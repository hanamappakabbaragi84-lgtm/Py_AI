```python
import os
import urllib.request
from pypdf import PdfReader

pdf_name = "sample.pdf"
url = "https://www.w3.org/WAI/ER/tests/xhtml/testfiles/resources/pdf/dummy.pdf"

# Ensure the PDF is downloaded
if not os.path.exists(pdf_name):
    print(f"Downloading sample PDF from {url}...")
    urllib.request.urlretrieve(url, pdf_name)
else:
    print(f"Using existing PDF: {pdf_name}")

# Extract text
reader = PdfReader(pdf_name)
text = ""
for page in reader.pages:
    text += page.extract_text() or ""

print("\nPDF processed successfully!")
print(f"Total length of extracted text: {len(text)} characters.")
```

```python
chunk_size = 700
overlap = 100

# Ensure text is defined in the current namespace
if 'text' not in globals():
    print("Warning: 'text' variable not found. Initializing text with an empty string.")
    text = ""

chunks = [
    text[i:i + chunk_size]
    for i in range(0, len(text), chunk_size - overlap)
    if text[i:i + chunk_size].strip()
]

print("Total chunks:", len(chunks))
if chunks:
    print("Preview of Chunk 1:", chunks[0])
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer(stop_words="english")
vectors = vectorizer.fit_transform(chunks)

def retrieve(question):
    query_vector = vectorizer.transform([question])
    scores = cosine_similarity(query_vector, vectors)[0]

    best_indices = scores.argsort()[-3:][::-1]

    return "\n".join(chunks[i] for i in best_indices)
```

```python
query = "What is in the PDF?"
retrieved_context = retrieve(query)
print(f"Query: {query}")
print(f"Retrieved Context: {retrieved_context}")
```
